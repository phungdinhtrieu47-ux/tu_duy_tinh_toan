bai 1 : [Cylinder] 
Xây dựng lớp Cylinder (hình trụ) gồm: bán kính và chiều cao 
(kiểu số thực) với hai phương thức:tính diện tích bề mặt (diện 
tích toàn phần) và thể tích của hình trụ 
Đầu vào  
Đâu vào từ bàn phím gồm một dòng duy nhất chứa bán kính và 
chiều cao của hình trụ, cách nhau bởi một dấu cách. 
 
Đầu ra 
In ra màn hình diện tích bề mặt và thể tích của hình trụ (sử 
dụng lại những hàm vừa viết, π=3.14).  
 
(Kết quả làm tròn đến 2 chữ số sau dấu phẩy). 

```python
class Cylinder:
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height
        self.pi = 3.14

    def surface_area(self):
        return 2 * self.pi * self.radius * (self.height + self.radius)

    def volume(self):
        return self.pi * (self.radius ** 2) * self.height
radius, height = map(float, input().split())
cylinder = Cylinder(radius, height)
area = cylinder.surface_area()
volume = cylinder.volume()
print(f"{area:.2f} {volume:.2f}")
```

Nhập vào một ngày tháng năm, (ngày tháng năm biểu diễn 
dưới dạng dd/mm/yyyy). Nếu ngày tháng năm không tồn tại, in 
ra dòng chữ INVALID. Nếu là hợp lệ in ra ngày kế tiếp của nó. 
 
Đầu vào  
Đầu vào từ bàn phim chứa dữ liệu ngày, tháng, năm dưới dạng 
dd/mm/yyyy 
. 
Đầu ra 
In ra màn hình ngày kế tiếp của ngày vừa nhập dưới dạng 
dd/mm/yyyy 
. Nếu ngày tháng không tồn tại, in ra INVALID 

```python
class MyDate:
    def __init__(self, day, month, year):
        self.dd = day
        self.mm = month
        self.yy = year
        
    def checkValid(self):
        maxday = -1
        match self.mm:
            case 1 | 3 | 5 | 7 | 8 | 10 | 12:
                maxday = 31
            case 4 | 6 | 9 | 11:
                maxday = 30
            case 2:
                if self.yy % 4 == 0 and (self.yy % 100 != 0 or self.yy % 400 == 0):
                    maxday = 29
                else:
                    maxday = 28
        return 1 <= self.dd <= maxday and 1 <= self.mm <= 12 and self.yy >= 0
    def getnextday(self):
        nextDay = MyDate(self.dd + 1, self.mm, self.yy)
        if nextDay.checkValid():
            return nextDay
        nextDay.dd = 1
        nextDay.mm += 1
        if nextDay.checkValid():
            return nextDay
        nextDay.mm = 1
        nextDay.yy += 1
        return nextDay
    def __str__(self):
        return f"{self.dd:02d}/{self.mm:02d}/{self.yy:04d}"
    d,m,y = map(int, input().split("/"))
    mydate1 = MyDate(d, m, y)
    print(mydate1.checkValid())
    print(mydate1)
    if mydate1.checkValid():
        nextday = mydate1.getnextday()
        print(nextday)
        
    
```

```python
class Calculator:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def set_numbers(self, a, b):
        self.a = a
        self.b = b

    def add(self):
        return self.a + self.b

    def subtract(self):
        return self.a - self.b

    def product(self):
        return self.a * self.b

    def divide(self):
        if self.b == 0:
            return "Lỗi: chia cho 0"
        return self.a / self.b

    def power(self):
        return self.a ** self.b

    def mod(self):
        if self.b == 0:
            return "Lỗi: chia cho 0"
        return self.a % self.b
a, b = map(float, input("Nhập 2 số ban đầu: ").split())
calc = Calculator(a, b)

while True:
    print("\n--- MENU ---")
    print("1. Cộng")
    print("2. Trừ")
    print("3. Nhân")
    print("4. Chia")
    print("5. Hàm mũ")
    print("6. Lấy phần dư")
    print("7. Nhập 2 số mới")

    choice = input("Chọn phép toán (1-7): ")

    if choice == "1":
        print("Kết quả:", calc.add())
    elif choice == "2":
        print("Kết quả:", calc.subtract())
    elif choice == "3":
        print("Kết quả:", calc.product())
    elif choice == "4":
        print("Kết quả:", calc.divide())
    elif choice == "5":
        print("Kết quả:", calc.power())
    elif choice == "6":
        print("Kết quả:", calc.mod())
    elif choice == "7":
        a, b = map(float, input("Nhập 2 số mới: ").split())
        calc.set_numbers(a, b)
    else:
        print("Lựa chọn không hợp lệ!")

    exit_choice = input("Bạn có muốn thoát? (yes/YES/Yes để thoát): ")
    if exit_choice in ["yes", "YES", "Yes"]:
        print("Chương trình kết thúc.")
        break
```

Với các phương thức sau: 
1. Khởi tạo với danh sách rỗng 
2. Thêm vào giỏ hàng một sản phẩm mới 
3. Xóa khỏi giỏ một sản phẩm nào đó (tìm theo tên để xóa) 
4. Kiểm tra xem giỏ hàng có rỗng không

```python
class ShoppingCart:
    def __init__(self):
        self.items = {}
    def add_item(self, name, price):
        if name in self.items:
            self.items[name] += price
        else:
            self.items[name] = price
    def remove_item(self, name):
        if name in self.items:
            del self.items[name]
    def giohangrong(self):
        return len(self.items) == 0
    def tongtien(self):
        return sum(self.items.values())
    def __str__(self):
        if self.giohangrong():
            return "Giỏ hàng trống"
        result = "Giỏ hàng:\n"
        for name, price in self.items.items():
            result += f"- {name}: {price:.2f}\n"
        return result
    def xoagiohang(self):
        self.items.clear()
cart = ShoppingCart()
while True:
    print("\n--- MENU ---")
    print("2. Thêm sản phẩm")
    print("3. Xóa sản phẩm")
    print("4. Kiểm tra giỏ hàng")
    print("5. Tính tổng tiền")
    print("6. Hiển thị giỏ hàng")
    print("7. Xóa giỏ hàng")

    choice = input("Chọn chức năng (2-7): ")

    if choice == "2":
        name = input("Nhập tên sản phẩm: ")
        price = float(input("Nhập giá sản phẩm: "))
        cart.add_item(name, price)
        print(f"Đã thêm {name} vào giỏ hàng.")
    elif choice == "3":
        name = input("Nhập tên sản phẩm cần xóa: ")
        cart.remove_item(name)
        print(f"Đã xóa {name} khỏi giỏ hàng.")
    elif choice == "4":
        if cart.giohangrong():
            print("Giỏ hàng trống.")
        else:
            print("Giỏ hàng có sản phẩm.")
    elif choice == "5":
        print(f"Tổng tiền: {cart.tongtien():.2f}")
    elif choice == "6":
        print(cart)
    elif choice == "7":
        cart.xoagiohang()
        print("Đã xóa tất cả sản phẩm khỏi giỏ hàng.")
    else:
        print("Lựa chọn không hợp lệ!")

    exit_choice = input("Bạn có muốn thoát? (yes/YES/Yes để thoát): ")
    if exit_choice in ["yes", "YES", "Yes"]:
        print("Chương trình kết thúc.")
        break
        
    
        
    
```

bai 6 :Yêu cầu 
Xây dựng lớp Rectangle(w, h) với các phương thức: 
 
area() trả về diện tích. 
 
perimeter() trả về chu vi. 
 
scale(k) phóng to hình chữ nhật theo hệ số k (nhân cả w và h), 
không trả về gì. 
 
Input 
w h k (số thực dương) 
 
Output 
area perimeter sau khi phóng to (làm tròn 2 chữ số thập phân).

```python
class Rectangle:
    def __init__(self, width, height):
        self.w = width
        self.h = height
    def area(self):
        return self.w * self.h
    def perimeter(self):
        return 2 * (self.w + self.h)
    def scale(self, k):
        self.w *= k
        self.h *= k
    def __str__(self):
        return f"Rectangle(width={self.w}, height={self.h})"
w, h, k = map(float, input("Nhập chiều rộng, chiều cao và hệ số tỷ lệ: ").split())
rect = Rectangle(w, h)
rect.scale(k)
print(rect)
print(rect.area())
print(rect.perimeter())
```

bai 7 : Yêu cầu 
Lớp Fraction(num, den) (tử, mẫu nguyên, mẫu ≠ 0) với: 
 
Rút gọn phân số trong __init__ (quy ước mẫu > 0). 
 
Các phương thức add/sub/mul/div trả về phân số đã rút gọn. 
 
__str__() in dạng a/b. 
 
Input 
a b op c d (tương ứng a/b (op) c/d, op thuộc {+,-,*,/}) 
 
Output 
Phân số kết quả đã rút gọn

```python
def uocchung(a, b):
    p = []
    if a > b:
        for i in range(1,b + 1):
            if a % i == 0 and b % i == 0:
                p.append(i)
    else:
        for i in range(1,a + 1):
            if a % i == 0 and b % i == 0:
                p.append(i)
    return max(p)
class Fraction:
    def __init__(self, tu, mau):
        self.tu = tu // uocchung(tu, mau)
        self.mau = mau // uocchung(tu, mau)
    def __add__(self, other):
        new_tu = self.tu * other.mau + other.tu * self.mau
        new_mau = self.mau * other.mau
        return Fraction(new_tu, new_mau)
    def __sub__(self, other):
        new_tu = self.tu * other.mau - other.tu * self.mau
        new_mau = self.mau * other.mau
        return Fraction(new_tu, new_mau)
    def __mul__(self, other):
        new_tu = self.tu * other.tu
        new_mau = self.mau * other.mau
        return Fraction(new_tu, new_mau)
    def __truediv__(self, other):
        new_tu = self.tu * other.mau
        new_mau = self.mau * other.tu
        return Fraction(new_tu, new_mau)
    def __str__(self):
        return f"{self.tu}/{self.mau}"
a, b = map(int, input("Nhập tử số và mẫu số của phân số 1: ").split())
c, d = map(int, input("Nhập tử số và mẫu số của phân số 2: ").split())
op = input("Nhập phép toán (+, -, *, /): ")
frac1 = Fraction(a, b)
frac2 = Fraction(c, d)
if op == "+":
    result = frac1 + frac2
elif op == "-":
    result = frac1 - frac2
elif op == "*":
    result = frac1 * frac2
elif op == "/":
    result = frac1 // frac2
else:
    result = "Phép toán không hợp lệ"
print("Kết quả:", result)
```

bai 8 : Yêu cầu
Lớp BankAccount(owner, balance) với:
deposit(x), withdraw(x); nếu rút quá số dư thì bỏ qua giao dịch.
__str__() in owner: balance.
Input
Dòng 1: owner initial_balance
Dòng 2: n (số lệnh)
Tiếp theo n dòng, mỗi dòng: DEPOSIT x hoặc WITHDRAW x.
Output
Số dư cuối cùng (in số thực, 2 chữ số).

```python
class BankAccount:
    def  __init__(self, owner, balance):
        self.o = owner
        self.b = balance
    def deposit(self, x):
        self.b += x
    def withdraw(self, x):
        if x < self.b:
            self.b -= x
        else:
            print("Số dư không đủ để rút")
    def __str__(self):
        return f"Tài khoản của {self.o} có số dư: {self.b:.2f}"
owner, balance = input("Nhập tên chủ tài khoản và số dư ban đầu: ").split()
balance = float(balance)
account = BankAccount(owner, balance)
n = int(input("Nhập số lần giao dịch: "))
for _ in range(n):
    action, amount = input("Nhập loại giao dịch (deposit/withdraw) và số tiền: ").split()
    amount = float(amount)
    if action == "deposit":
        account.deposit(amount)
    elif action == "withdraw":
        account.withdraw(amount)
    else:
        print("Loại giao dịch không hợp lệ")
print(account)
```

bai 9 : Yêu cầu
Lớp Student(name) với:
add_score(subject, score)
avg() trả về điểm trung bình.
rank() trả về "Excellent" nếu avg ≥ 8, "Good" nếu 6.5–<8,
"Average" nếu 5–<6.5, ngược lại "Poor".
Input
Dòng 1: name
Dòng 2: n
n dòng tiếp: subject score (score là số thực 0–10)
Output
name avg rank (avg 2 chữ số).

```python
class Student:
    def __init__(self, name):
        self.ten = name
        self.subjects = {}
    def add_score(self, subject, score):
        self.subjects[subject] = score
    def avg(self):
        if not self.subjects:
            return 0
        return sum(self.subjects.values()) / len(self.subjects)
    def rank(self):
        if self.avg() >= 8:
            return "Excellent"
        elif 8 > self.avg() >= 6.5:
            return "Good"
        else:
            return "Poor"
    def __str__(self):
        return f"Học sinh: {self.ten}, Điểm trung bình: {self.avg():.2f}, Xếp loại: {self.rank()}"
name = input("Nhập tên học sinh: ")
n = int(input("Nhập số môn học: "))
student = Student(name)
for _ in range(n):
    subject, score = input("Nhập tên môn học và điểm số: ").split()
    score = float(score)
    student.add_score(subject, score)
print(student)
```

bai 11 : Bạn được cho bốn điểm A, B, C và D trong hệ tọa độ Đề-các 3
chiều.
Nhiệm vụ của bạn là tính và in ra góc giữa hai mặt phẳng tạo
bởi các điểm:
mặt phẳng (A, B, C)
mặt phẳng (B, C, D)
Góc cần in ra được ký hiệu là PHI, và phải dùng độ (degrees),
không dùng radian.
Dạng Input:
Một dòng input chứa 3 số thực, ngăn cách bằng dấu cách, biểu
diễn:
tọa độ X tọa độ Y tọa độ Z của một điểm.
Dạng Output:
In ra góc giữa hai mặt phẳng, chính xác tới hai chữ số thập
phân. 

```python
import math
class Point:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z
    def vector(self, other):
        return (other.x - self.x, other.y - self.y, other.z - self.z)
A = map(float, input("Nhập tọa độ điểm A (x y z): ").split())
B = map(float, input("Nhập tọa độ điểm B (x y z): ").split())
C = map(float, input("Nhập tọa độ điểm C (x y z): ").split())
D = map(float, input("Nhập tọa độ điểm D (x y z): ").split())
pointA = Point(*A)
pointB = Point(*B)
pointC = Point(*C)
pointD = Point(*D)
vectorAB = pointA.vector(pointB)
vectorCD = pointC.vector(pointD)
vectorBC = pointB.vector(pointC)
def tichcohuong(v1, v2):
    return ((
        v1[1]*v2[2] - v1[2]*v2[1],
        v1[2]*v2[0] - v1[0]*v2[2],
        v1[0]*v2[1] - v1[1]*v2[0]
    ))
abc = tichcohuong(vectorAB, vectorBC)
bcd = tichcohuong(vectorBC, vectorCD)
goc = math.acos((abc[0]*bcd[0] + abc[1]*bcd[1] + abc[2]*bcd[2]) / (math.sqrt(abc[0]**2 + abc[1]**2 + abc[2]**2) * math.sqrt(bcd[0]**2 + bcd[1]**2 + bcd[2]**2)))
print(f"Góc giữa hai mặt phẳng ABC và BCD là: {math.degrees(goc):.2f} độ")
    
```

của một số phức và hàm ComplexNumber
multiply(complexNumber other) trả về tích của số phức hiện tại
và một số phức khác.
Đầu vào
Đầu vào từ bàn phím gồm hai dòng, mỗi dòng gồm hai số thực
cách nhau bởi một dấu cách biểu diễn phần thực và phần ảo
của một số phức. Các số trong đầu vào có giá trị tuyệt đối
không vượt quá \(100\).
Đầu ra
In ra màn hình hai số biểu diễn phần thực và phần ảo của tích
hai số phức.

```python
class ComplexNumber:
    def __init__(self, initRealPart, initImaginaryPart):
        self.realPart = initRealPart
        self.imaginaryPart = initImaginaryPart

    def multiply(self, other):
        real = self.realPart * other.realPart - self.imaginaryPart * other.imaginaryPart
        imag = self.realPart * other.imaginaryPart + self.imaginaryPart * other.realPart
        return ComplexNumber(real, imag)

r1, i1 = map(float, input().split())
r2, i2 = map(float, input().split())

c1 = ComplexNumber(r1, i1)
c2 = ComplexNumber(r2, i2)

result = c1.multiply(c2)

print(result.realPart, result.imaginaryPart)
```

bai 13 : [Triangle]
Khai báo cấu trúc dữ liệu tam giác \(Triangle\) chứa \(3\) biến
\(egde1,\,edge2,\,edge3\) kiểu \(int\) là độ dài \(3\) cạnh của
tam giác đó.
Viết hàm khởi tạo Triangle(int initEdge1,int initEdge2, int
initEdge3) để khởi tạo độ dài các cạnh của tam giác và hàm
double getArea() trả về diện tích của tam giác đó (nếu tam giác
không hợp lệ, trả về \(-1\)).
Hoàn thiện chương trình dưới đây nhập vào \(3\) cạnh của một
tam giác và in ra diện tích của tam giác đó.
Đầu vào
Một dòng duy nhất từ bàn phím chứa \(3\) số nguyên là độ dài
\(3\) cạnh của tam giác
Đầu ra
In ra màn hình diện tích tam giác (làm tròn đến\(2\) chữ số thập
phân sau dấu phẩy). In ra \(invalid\) nếu \(3\) số nhập vào
không phải là \(3\) cạnh của một tam giác 

```python
class Triangle:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    def is_valid(self):
        return self.a + self.b > self.c and self.a + self.c > self.b and self.b + self.c > self.a
    def area(self):
        if not self.is_valid():
            return -1
        s = (self.a + self.b + self.c) / 2
        return (s * (s - self.a) * (s - self.b) * (s - self.c)) ** 0.5
a, b, c = map(float, input("Nhập độ dài 3 cạnh của tam giác: ").split())
triangle = Triangle(a, b, c)
if triangle.is_valid():
    print(f"Diện tích tam giác: {triangle.area():.2f}")
else:
    print("Không phải là tam giác hợp lệ")
```

bai 15 : Yêu cầu
Xây dựng lớp Employee(name, salary).
Xây dựng lớp Manager(name, salary, bonus) kế thừa từ
Employee.
Cả hai lớp đều có phương thức income():
- Employee: trả về salary
- Manager: trả về salary + bonus
Input
Dòng 1: loại đối tượng (Employee hoặc Manager)
Dòng 2: name salary [bonus]
Output
In ra tổng thu nhập của đối tượng. 


```python
class Emloyee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def __str__(self):
        return f"Nhân viên: {self.name}, Lương: {self.salary:.2f}"
    def thunhập(self):
        return self.salary
class Manager:
    def __init__(self, name, salary, bonus):
        self.name = name
        self.salary = salary
        self.bonus = bonus
    def __str__(self):
        return f"Quản lý: {self.name}, Lương: {self.salary:.2f}, Thưởng: {self.bonus:.2f}"
    def total_compensation(self):
        return self.salary + self.bonus
doituong = input("Nhập loại đối tượng (employee/manager): ")
if doituong == "employee":
    name = input("Nhập tên nhân viên: ")
    salary = float(input("Nhập lương nhân viên: "))
    employee = Emloyee(name, salary)
    print(employee)
    print(f"Thu nhập của nhân viên: {employee.thunhập():.2f}")
elif doituong == "manager":
    name = input("Nhập tên quản lý: ")
    salary = float(input("Nhập lương quản lý: "))
    bonus = float(input("Nhập thưởng của quản lý: "))
    manager = Manager(name, salary, bonus)
    print(manager)
    print(f"Tổng thu nhập của quản lý: {manager.total_compensation():.2f}")
```
