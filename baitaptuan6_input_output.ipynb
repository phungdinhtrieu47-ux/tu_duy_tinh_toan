W6A4: Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo
dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key lặp
lại, thêm vào list của key đó 

```python
a = input()
d = {}
b = a.split()
for x in b:
    k, v = x.split(':')
    if k not in d:
        d[k] = []
    d[k].append(v)
print(d)    
```

W6A5: Nhập một list số nguyên, In ra dictionary {'positives': ..., 'negatives': ..., 'zeros':
...} (đếm số >0, <0, =0)

```python
a = list(map(int, input().split()))

d = {
    'positives': 0,
    'negatives': 0,
    'zeros': 0
}

for x in a:
    if x > 0:
        d['positives'] += 1
    elif x < 0:
        d['negatives'] += 1
    else:
        d['zeros'] += 1

print(d)
```

W6A9: Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một dictionary mới
trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ được cộng lại, nếu chỉ có
trong một dict thì giữ nguyên. In ra dictionary mới theo thứ tự từ điển tăng dần của
key. 

```python
d1 = eval(input())
d2 = eval(input())

d = {}

for k in d1:
    d[k] = d1[k]

for k in d2:
    if k in d:
        d[k] += d2[k]
    else:
        d[k] = d2[k]

res = {}

for k in sorted(d):
    res[k] = d[k]

print(res)
```

W6A10: Viết chương trình nhập vào một danh sách các số nguyên, và số nguyên dương k. In ra list gồm các cặp số khác nhau có tổng bằng k.
Mỗi cặp là một tuple, hai số trong tuple có thể giống nhau, thứ tự xuất hiện trong danh sách của số thứ nhất nhỏ hơn thứ tự của số thứ hai.
Các tuple trong list được sắp xếp theo thứ tự tăng dần của phần tử đầu tiên trong tuple.

```python
nums = list(map(int, input().split()))
k = int(input())

res = []

for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        if nums[i] + nums[j] == k:
            res.append((nums[i], nums[j]))

res.sort()

print(res)
```

W6A11: Nhập vào một tuple chứa các số nguyên. Hãy tạo ra hai tuple mới:

một chứa tất cả các số chẵn
một chứa tất cả các số lẻ từ tuple ban đầu

Giữ nguyên thứ tự tương đối. In ra hai tuple này trên hai dòng riêng biệt.

```python
t = tuple(map(int, input().split()))

chan = ()
le = ()

for x in t:
    if x % 2 == 0:
        chan += (x,)
    else:
        le += (x,)

print(chan)
print(le)
```

W6A12: Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong list.
Nếu có nhiều số cùng có số lần xuất hiện nhiều nhất, hãy in ra số nhỏ nhất trong số đó.

```python
nums = list(map(int, input().split()))

d = {}

for x in nums:
    if x in d:
        d[x] += 1
    else:
        d[x] = 1

max_count = max(d.values())

res = None

for key in d:
    if d[key] == max_count:
        if res is None or key < res:
            res = key

print(res)
```

W6A13: Nhập vào một dictionary có value là duy nhất (không trùng nhau).
Hãy tạo một dictionary mới bằng cách đảo ngược key và value của dictionary ban đầu. In ra dictionary mới.

```python
d = eval(input())

new_d = {}

for key in d:
    value = d[key]
    new_d[value] = key

print(new_d)
```

W6A14: Nhập vào hai dòng, mỗi dòng là một danh sách các số nguyên.
Hãy tìm các phần tử chung (giao) của hai danh sách này và in ra một list mới chứa các phần tử chung đó, không trùng lặp.

```python
a = list(map(int, input().split()))
b = list(map(int, input().split()))

res = []

for x in a:
    if x in b and x not in res:
        res.append(x)

print(res)
```

W6A15: Nhập vào một dictionary có key là chuỗi và value là số nguyên, và một số nguyên k ở dòng tiếp theo.
Hãy tạo một dictionary mới chỉ chứa các cặp key-value từ dictionary ban đầu mà có value lớn hơn k.
In ra dictionary mới.

```python
d = eval(input())
k = int(input())

new_d = {}

for key in d:
    if d[key] > k:
        new_d[key] = d[key]

print(new_d)
```
