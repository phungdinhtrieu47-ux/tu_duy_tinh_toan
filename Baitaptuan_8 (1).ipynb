Xây dựng lớp Rectangle(w, h) với các phương thức:
area() trả về diện tích.
perimeter() trả về chu vi.
scale(k) phóng to hình chữ nhật theo hệ số k (nhân cả w và h),
không trả về gì.
Input
w h k (số thực dương)
Output
area perimeter sau khi phóng to (làm tròn 2 chữ số thập phân).

```python
class Rectangle:
    def __init__(self,w,h):
        self.w = h
        self.h = h
    def area(self):
        return self.h * self.w
    def perimeter(self):
        return (self.h + self.w) * 2
    def scale(self, k)
        self.w *= k
        self.h *= k

w, h, k = map(int, input().split())
rect = Rectangle(w, h)
rect.scale(k)
print(f"{rect.area():.2f}")
print(f"{rect.perimeter():.2f}")    
```

Lớp Fraction(num, den) (tử, mẫu nguyên, mẫu ≠ 0) với:
Rút gọn phân số trong __init__ (quy ước mẫu > 0).
Các phương thức add/sub/mul/div trả về phân số đã rút gọn.
__str__() in dạng a/b.
Input
a b op c d (tương ứng a/b (op) c/d, op thuộc {+,-,*,/})
Output
Phân số kết quả đã rút gọn 


```python
import math
class Fraction:
    def __init__(self, num, den):
        if den < 0:
            num = -num
            den = -den
        gcd = math.gcd(abs(num),den)    
        self.num = num // gcd
        self.den = den // gcd
    
    def add(self, other):
        new_num = self.num * other.den + self.den * other.num
        new_den = self.den * other.den
        return Fraction(new_num, new_den)
    
    def sub(self, other):
        new_num = self.num * other.den - self.den * other.num
        new_den = self.den * other.den
        return Fraction(new_num, new_den)
    
    def mul(self, other):
        new_num = self.num * other.num
        new_den = self.den * other.den
        return Fraction(new_num, new_den)
    
    def div(self, other):
        new_num = self.num * other.den
        new_den = self.den * other.num
        return Fraction(new_num, new_den)
    
    def __str__(self):
        return f"{self.num}/{self.den}"
    
a, b, op, c, d = input().split()
a = int(a)
b = int(b)
c = int(c)
d = int(d)
f1 = Fraction(a, b)
f2 = Fraction(c, d)
if op == '+':
    result = f1.add(f2)
elif op == '-':
    result = f1.sub(f2)
elif op == '*':
    result = f1.mul(f2)
elif op == '/':
    result = f1.div(f2)
else:
    print("Toán tử không hợp lệ")
    exit()

print(result)    
```

Lớp BankAccount(owner, balance) với:
deposit(x), withdraw(x); nếu rút quá số dư thì bỏ qua giao dịch.
__str__() in owner: balance.
Input
Dòng 1: owner initial_balance
Dòng 2: n (số lệnh)
Tiếp theo n dòng, mỗi dòng: DEPOSIT x hoặc WITHDRAW x.
Output
Số dư cuối cùng (in số thực, 2 chữ số).

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
    def deposit(self,x):
        if x > 0:
            self.balance += x
    def withdraw(self, x):
        if x > 0 and x < self.balance:
            self.balance -= x
    def __str__(self):
        return f"{self.owner}: {self.balance}"        
owner, initial_balance = input().split()
initial_balance = float(initial_balance)
account = BankAccount(owner, initial_balance)
n = int(input())
for _ in range(n):
    command = input().split()
    action = command[0]
    amount = float(command[1])
    if action == "DEPOSIT":
        account.deposit(amount)
    elif action == "WITHDRAW":
        account.withdraw(amount)

print(account)        
```

Lớp Student(name) với:
add_score(subject, score)
avg() trả về điểm trung bình.
rank() trả về "Excellent" nếu avg ≥ 8, "Good" nếu 6.5–<8,
"Average" nếu 5–<6.5, ngược lại "Poor".
Input
Dòng 1: name
Dòng 2: n
n dòng tiếp: subject score (score là số thực 0–10)
Output
name avg rank (avg 2 chữ số). 

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.score = []
    def add_score(self, subject, score):
        if 0 <= score <= 10:
            self.score.append(score)
    def avg(self):
        if len(self.score) == 0:
            return 0
        return sum(self.score) / len(self.score)
    def rank(self):
        average = self.avg()
        if average >= 8:
            return "Excellent"
        elif average >= 6.5:
            return "Good"
        elif average >= 5:
            return "Average"
        else:
            return "Poor"
name = input()
n = int(input())
student = Student(name)
for _ in range(n):
    part = input().strip().split()
    subject = part[0]
    score = float(part[1])
    student.add_score(subject, score) 
average = student.avg()
rank_result = student.rank()

print(f"{student.name} {average:.2f} {rank_result}")                       
```

Lớp Book(title, author, year).
Lớp Library() với:
add(book)
count_by_author(name)
find_by_year(y) trả về số sách xuất bản năm y.
Input
Dòng 1: n (số thao tác)
n dòng:
ADD title;author;year
COUNT author (đếm theo tác giả)
COUNTYEAR year (đếm theo năm)
Output
Mỗi lệnh COUNT/COUNTYEAR in một dòng kết quả.

```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
class library:
    def __init__(self):
        self.books = []
    def add(self, book):
        self.books.append(book)
    def count_by_author(self, name):
        count = 0
        for book in self.books:
            if book.author == name:
                count += 1
        return count
    def find_by_year(self, y):
        count = 0
        for book in self.books:
            if book.year == y:
                count += 1
        return count
n = int(input().strip())

library = Library()

for _ in range(n):
    command = input().strip()
    
    if command.startswith("ADD"):
        # Định dạng: ADD title;author;year
        parts = command[4:].split(";")  # Bỏ "ADD " ở đầu
        title = parts[0].strip()
        author = parts[1].strip()
        year = int(parts[2].strip())
        
        book = Book(title, author, year)
        library.add(book)
    
    elif command.startswith("COUNT"):
        # Định dạng: COUNT author
        author_name = command[6:].strip()  # Bỏ "COUNT " ở đầu
        result = library.count_by_author(author_name)
        print(result)
    
    elif command.startswith("COUNTYEAR"):
        # Định dạng: COUNTYEAR year
        year = int(command[10:].strip())  # Bỏ "COUNTYEAR " ở đầu
        result = library.find_by_year(year)
        print(result)                         
```
